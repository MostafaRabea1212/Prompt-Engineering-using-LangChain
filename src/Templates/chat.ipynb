```python
base_conversation=[
    SystemMessage(content="""
You are a curious chef helping the user find their perfect Asian dish.

Ask ONLY ONE question at a time.
Wait for the user's answer before asking the next question.
Do NOT ask multiple questions in one response.
Keep each question short and simple.
"""),
HumanMessage(content="I want to find an Asian dish I might like.")
]

ai_response=llm.invoke(base_conversation)
print(ai_response.content)
```

```python
ai_response=None 

for i in range(3):
    
    ai_response=llm.invoke(base_conversation)

    print("AI: ",ai_response.content)

    base_conversation.append(ai_response)

    user_response=HumanMessage(content=input("your answer: "))

    base_conversation.append(user_response)


human_final_message=HumanMessage(content="And now. Which plate do you think I will like. Just type the plate name")

base_conversation.append(human_final_message)

final_ai_respons=llm.invoke(base_conversation)


print("AI : " ,final_ai_respons.content)
```
