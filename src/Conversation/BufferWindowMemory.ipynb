```python
import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..", "..")))

from src.Generate.Generate_By_Api import BaseLLM

llm = BaseLLM()
```

```python
from langchain.agents import create_agent, AgentState
from langchain.agents.middleware import before_model
from langchain.messages import RemoveMessage
from langchain_core.callbacks import UsageMetadataCallbackHandler
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.graph.message import REMOVE_ALL_MESSAGES
from langgraph.runtime import Runtime
from typing import Any
```

```python
cb=UsageMetadataCallbackHandler()
```

### Bufffer Window Memory

```python
message_1 = "Can you list 3 places in Paris to visit?"

message_2 = "Which place has a lake view?"

message_3 = "Can I visit in winter?"
```

```python
HISTORY_WINDOW = 4
```

```python
@before_model
def keep_window(state: AgentState, runtime: Runtime) -> dict[str, Any] | None:
    messages = state["messages"]
    limit = HISTORY_WINDOW + 1
    if len(messages) <= limit:
        return None
    return {"messages": [RemoveMessage(id=REMOVE_ALL_MESSAGES), *messages[-limit:]]}


agent= create_agent(
    model=llm.llm,
    system_prompt="you are a helpful assistant",
    middleware=[keep_window],
    checkpointer=InMemorySaver(),
)
```

```python

output_1=agent.invoke(
    {
        "messages" : [
            {
                "role" : "user",
                "content" : message_1
            }
        ]
    },
    config={
        "configurable" : {"thread_id" : "user_1"},
        "callbacks" : [cb]
    }

)
```

```python
print(output_1["messages"][-1].content) 
```

```python
print("======================")
print(cb.usage_metadata)
```
