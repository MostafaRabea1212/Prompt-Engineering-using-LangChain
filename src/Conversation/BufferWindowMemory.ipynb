```python
import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..", "..")))

from src.Generate.Generate_By_Api import BaseLLM
from langchain_core.prompts import PromptTemplate

llm = BaseLLM()
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.callbacks import UsageMetadataCallbackHandler
```

```python
cb=UsageMetadataCallbackHandler()
```

### Bufffer Window Memory

```python
prompt=ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistant."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])


chain= prompt | llm.llm

store= {}

def get_session_history(session_id : str):

    if session_id not in store:

        store[session_id]=InMemoryChatMessageHistory()

    history= store[session_id]

    if len(history.messages) > 2 : 
        history.messages=history.messages[-2 : ]
        
    print(f"[{session_id}] history length AFTER trim: {len(history.messages)}")

    return history
```

```python
conversation=RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history"
)
```

```python
message_1 = "Can you list 3 places in Paris to visit?"

message_2 = "Which place has a lake view?"

message_3 = "Can I visit in winter?"
```

```python
output_1=conversation.invoke(
    {"input" : message_1},
    config={"configurable" : {"session_id" : "user_1"}
            ,"callbacks":[cb]}
)
print(output_1.content)
```
